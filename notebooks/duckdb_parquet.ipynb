# Query Parquet Exports With DuckDB

Download the Actor's `PARQUET_*` key-value-store files, then point DuckDB at the local files.

```python
import duckdb

con = duckdb.connect()
con.sql("""
    select buyer_country, count(*) as notices
    from read_parquet('data/notices.parquet')
    group by buyer_country
    order by notices desc
    limit 20
""").df()
```

```python
con.sql("""
    select main_cpv, count(*) as notices, sum(cast(estimated_value as double)) as total_value
    from read_parquet('data/notices.parquet')
    group by main_cpv
    order by notices desc
    limit 20
""").df()
```
