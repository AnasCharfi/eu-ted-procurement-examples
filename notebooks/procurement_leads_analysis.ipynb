Use this notebook after exporting a run dataset from Apify as CSV or JSON.

```python
import pandas as pd

df = pd.read_csv('../crm/sample_crm_import.csv')
df.head()
```

```python
df.groupby('buyer_country').agg(notices=('notice_id', 'count'), total_value=('estimated_value', 'sum'))
```

```python
df.assign(cpv_prefix=df['main_cpv'].astype(str).str[:2]).groupby('cpv_prefix')['notice_id'].count().sort_values(ascending=False)
```
